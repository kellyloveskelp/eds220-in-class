```python
import pandas as pd
import numpy as np
```

```python
# A numpy array
arr = np.random.randn(4) # random values from std normal distribution
print(type(arr))
print(arr, "\n")

# A pandas series made from the previous array
s = pd.Series(arr)
print(type(s))
print(s)
```

```python
# The basic method to create a pandas.Series is to call
# s = pd.Series(data, index=index)
```

```python
# A series from a numpy array 
pd.Series(np.arange(3), index=[2023, 2024, 2025])
```

```python
# A series from a list of strings with default index
pd.Series(['EDS 220', 'EDS 222', 'EDS 223', 'EDS 242'])
```

```python
# Construct dictionary
d = {'key_0':2, 'key_1':'3', 'key_2':5}

# Initialize series using a dictionary
pd.Series(d)
```

```python
pd.Series(3.0, index = ['A', 'B', 'C'])
```

```python
# Define a series
s = pd.Series([98,73,65],index=['Andrea', 'Beth', 'Carolina'])

# Divide each element in series by 10
print(s /10, '\n')

# Take the exponential of each element in series
print(np.exp(s), '\n')

# Original series is unchanged
print(s)
```

```python
s > 70
```

```python
# Series with NAs in it
s = pd.Series([1, 2, np.nan, 4, np.nan])
s
```

```python
# Check if series has NAs
s.hasnans
```

```python
s.isna()
```

```python
# Check in:
# The integer number -999 is often used to represent missing values. 
# Create a pandas.Series named s with four integer values, two of which are -999. 
# The index of this series should be the letters A through D.
```

```python
s = pd.Series([-999,73, -999, 65],index=['A', 'B', 'C', 'D'])
s
```

```python
# Check in:
# In the pandas.Series documentation, look for the method mask(). 
# Use this method to update the series s so that the -999 values are 
# replaced by NA values. 
# HINT: check the first example in the method’s documentation. 
```

```python
s_masked = s.mask(s == -999) # mask() assigns NaN to -999
s_masked
```

```python
# Initialize dictionary with columns' data 
d = {'col_name_1' : pd.Series(np.arange(3)), # assigns variables to values 0-2
     'col_name_2' : pd.Series([3.1, 3.2, 3.3]), # sets variables as defined
     }

# Create data frame
df = pd.DataFrame(d)
df
```

```python
# Change index
df.index = ['a','b','c']
df
```

```python
# Check in: We can access the data frame’s column names via the columns attribute. 
# Update the column names to C1 and C2 by updating this attribute.
```

```python
df.columns = ['C1', 'C2']
df
```
